# No-Show Baseline Model — Logistic Regression

Goal: train and evaluate a baseline logistic regression model on the cleaned no-show data.

## Load cleaned data

```python
import pandas as pd

model_df = pd.read_csv('../data/processed/no_show_clean.csv')
model_df.shape
```

```python
model_df.columns
```

## Split into features (X) and target (y)

```python
X = model_df.drop(columns = ['PatientId', 'AppointmentID', 'ScheduledDay', 'AppointmentDay', 'No-show', 'no_show_flag'])
y = model_df['no_show_flag']
X.columns
```

## One-hot encode categorical columns

```python
X_encoded = pd.get_dummies(X,drop_first = True)
X_encoded.shape
```

```python
X_encoded.columns
```

## Train/test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size = 0.2, stratify = y, random_state = 42)
X_train.shape , X_test.shape
```

## Train baseline logistic regression

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter = 1000)
model.fit(X_train, y_train)
```

## Evaluate

```python
from sklearn.metrics import classification_report, roc_auc_score
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:,1]

print(classification_report(y_test,y_pred))
print('ROC-AUC:', roc_auc_score(y_test,y_proba))
```

## Retrain with class_weight='balanced'

```python
model_balanced = LogisticRegression(max_iter=1000, class_weight='balanced')
model_balanced.fit(X_train, y_train)

y_pred_bal = model_balanced.predict(X_test)
y_proba_bal = model_balanced.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred_bal))
print("ROC-AUC:", roc_auc_score(y_test,y_proba_bal))
```

## Threshold tuning

```python
from sklearn.metrics import precision_score, recall_score, f1_score
thresholds = [0.3, 0.4, 0.5, 0.6, 0.7]
for t  in thresholds:
       y_pred_t = (y_proba_bal >= t).astype(int)
       precision = precision_score(y_test,y_pred_t)
       recall = recall_score(y_test, y_pred_t)
       f1 = f1_score(y_test, y_pred_t)
       print(f"threshold={t}: precision={precision:.2f}, recall={recall:.2f}, f1={f1:.2f}")
```

## Inspect learned feature weights

```python
coefs  = pd.Series(model_balanced.coef_[0], index = X_encoded.columns)
coefs.sort_values()
```

```python
print(coefs.head(10))
print(coefs.tail(10))
print(coefs['wait_days'])
```

```python
coefs = coefs.sort_values()
print(coefs.head(10))
print(coefs.tail(10))
```

## Save the model

```python
import joblib
joblib.dump(model_balanced,'../models/no_show/logistic_regression_v1.joblib')
```

## Gradient boosting comparison

```python
from sklearn.ensemble import HistGradientBoostingClassifier

gb_model = HistGradientBoostingClassifier(class_weight = 'balanced', random_state = 42)
gb_model.fit(X_train, y_train)

y_pred_gb = gb_model.predict(X_test)
y_proba_gb = gb_model.predict_proba(X_test)[:,1]

print(classification_report(y_test,y_pred_gb))
print(roc_auc_score(y_test,y_proba_gb))
```

### Threshold tuning for gradient boosting

```python
for t in thresholds:
  y_pred_t = (y_proba_gb >=t).astype(int)
  precision = precision_score(y_test, y_pred_t)
  recall = recall_score(y_test, y_pred_t)
  f1 = f1_score(y_test, y_pred_t)
  print(f"threshold={t}: precision={precision:.2f}, recall={recall:.2f}, f1={f1:.2f}")
```

## Permutation importance for gradient boosting

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(gb_model, X_test, y_test, n_repeats = 10, random_state = 42, scoring = 'roc_auc')
importances = pd.Series(result.importances_mean, index= X_test.columns)
importances.sort_values(ascending=False).head(15)
```
